### Detecting horizontal lines

```python
def detect_horizontal_lines(image, threshold=1):
    image = image.unsqueeze(0)

    kernel = torch.tensor([
        [-1, -1, -1],
        [ 2,  2,  2],
        [-1, -1, -1]
    ], dtype=torch.float32).unsqueeze(0).unsqueeze(0)

    kernel = kernel.to(device=image.device, dtype=image.dtype)
    response = F.conv2d(image, kernel, padding=1)

    positions = []
    number = 0

    return response

res = detect_horizontal_lines(patch[100])

plt.imshow(torch.clamp(res[0,0,:,:], min=0))
plt.colorbar()
```
